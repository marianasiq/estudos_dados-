```python
import pandas as pd
```

```python
df = pd.read_csv("vendas_suja.csv")
df.head(10)
df.info()
```

```python
df.isnull().sum() # é usado para encontrar nulos e somar
```

```python
df[df["preco"].isnull()] # para ver em preço qual o valor nulo
```

```python
'''
Poucos nulos e a coluna é essencial	Remover a linha: df.dropna(subset=["preco"])
Coluna de texto sem informação	Preencher com um rótulo: df["cidade"].fillna("Não informada")
Coluna numérica	Preencher com média ou mediana: df["quantidade"].fillna(df["quantidade"].median())'''
```

```python
df.dropna(subset=["preco"]) # apagou essa linha pois não da para seguir sem o valor do preço
```

```python
df["cidade"] = df["cidade"].fillna("Não informada") # o fillna serve para preencher o espaço vazio
df
```

```python
df["pagamento"] = df["pagamento"].fillna("Não informado")
df
```

```python
df["quantidade"] = df["quantidade"].fillna(df["quantidade"].median()) 
df
```

```python
df.duplicated().sum()        # somar quantas linhas são repetidas
df[df.duplicated(keep=False)]  # ver as duplicadas
```

```python
df = df.drop_duplicates() # apagar as linhas duplicadas
```

```python
#quando o Pandas lê o CSV, a coluna data vem como texto, e não como data!! então é necessario formatar
df["data"].dtype #aparece o valor de string
```

```python
df["data"] = pd.to_datetime(df["data"], format="mixed", dayfirst=True) #primeiro coverte a coluna pro formato data, afirna que o formato é misto e deixa claro que o dia vem primeiro(padrão brasileiro)
```

```python
df["data"].dtype
df["data"].head()
```

```python
df["mes"] = df["data"].dt.month
df["dia_semana"] = df["data"].dt.day_name()  #day_name traz nome dos dias da semana
df[["data", "mes", "dia_semana"]].head()
```

```python
df["quantidade"] = df["quantidade"].astype(int)  #converter a coluna de quantidade  pro tipo inteiro
```

```python
for i in ["produto", "categoria", "cidade", "pagamento"]:
    df[i] = df[i].str.strip().str.title() #.str pega as ferramentas de texto, .strip() remove os espaços, .title() deixa somente a primeira letra maiuscula
```

```python
df["produto"].value_counts() # são se repete mais
```

```python
df["quantidade"].describe() #describe mostra estatisticas da coluna 
```

```python
df[df["quantidade"] > 50] # como na maioria não passa de 50 então esse valor deve estar errado
```

```python
df = df[df["quantidade"] <= 50] #como os valores vão de um a seis então é melhor excluir essa linha ou poderia corrigir supondo que foi um erro de digitação: df.loc[df["quantidade"] == 500, "quantidade"] = 5
```

```python
df = df.dropna(subset=["preco"]) #remove as linhas em que preco está vazio
```

```python
df["receita"] = df["preco"] * df["quantidade"] # tabela de receita  
```

```python
df.to_csv("vendas_limpa.csv", index=False) #cria um csv com as alterações feitas
```
