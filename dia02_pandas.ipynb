```python
import pandas as pd
import numpy as np
```

```python
dados = {
    "produto": ["Camiseta", "Calça", "Tênis", "Boné"],
    "categoria": ["Roupa", "Roupa", "Calçado", "Acessório"],
    "preco": [50, 120, 250, 40],
    "quantidade": [10, 5, 3, 20],
}
data_frame = pd.DataFrame(dados)
data_frame #dataframe é uma tabela
```

```python
data_frame.head()    
```

```python
data_frame.shape #numero de linhas e colunas         
```

```python
data_frame.columns #nome das colunas
```

```python
data_frame.dtypes #tipo de variavel de cada coluna
```

```python
data_frame.info() #resumo geral da tabela
```

```python
data_frame.describe() # estatísticas das colunas numéricas
```

```python
data_frame[["preco", "quantidade"]]
data_frame["receita"] = data_frame["preco"] * data_frame["quantidade"]
data_frame
```

```python
data_frame[data_frame["preco"] > 100]                          
```

```python
data_frame[data_frame["categoria"] == "Roupa"]                
  
```

```python
data_frame[(data_frame["preco"] > 40) & (data_frame["quantidade"] > 5)]   
```

```python
data_frame[(data_frame["categoria"] == "Roupa") | (data_frame["preco"] > 200)]
```

```python
data_frame.sort_values("preco")                          # do menor para o maior
```

```python
data_frame.sort_values("receita", ascending=False)       # do maior para o menor
```

```python
# loc e iloc: iloc = por posição (números, como em listas). loc = por nome (rótulos e condições).
data_frame.iloc[0]
```

```python
data_frame.loc[data_frame["preco"] > 100, ["produto", "preco"]]
```
