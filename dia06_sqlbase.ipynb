```python
import sqlite3 
import pandas as pd
```

```python
conexão = sqlite3.connect("loja.db") # .connect faz a conexão entre o sql e o data base

def sql(consulta):
    return pd.read_sql_query(consulta, conexão)
sql("SELECT * FROM vendas LIMIT 5")
```

```python
sql("""
SELECT produto, preco
FROM vendas
WHERE preco > 100
""") # para instruções longas se ultiliza as 3 aspas
```

```python
sql ("SELECT produto, cidade, preco FROM vendas") # select pega as colunas e from informa de qual tabela
```

```python
sql ("""SELECT produto, preco, quantidade, preco * quantidade AS receita
FROM vendas""") # o AS da nome a novas colunas como essa calculada
```

```python
sql("SELECT DISTINCT cidade FROM vendas ORDER BY cidade") # distinct traz os valores sem repetições 
```

```python
sql("SELECT produto, preco FROM vendas WHERE preco > 190") # WHERE é usado para definir uma condição
```

```python
sql(""" SELECT produto, preco, quantidade, preco * quantidade AS receita
FROM vendas
WHERE categoria = 'Roupa' AND receita > 200
ORDER BY receita DESC """) # DESC é do maior valor ao menor
```

```python
sql("""SELECT id_venda, produto, preco FROM vendas
WHERE preco BETWEEN 50 AND 120""") # BETWEEN traz valores de 50 a 120
```

```python
sql("SELECT produto FROM vendas WHERE produto LIKE 'C%'") # o LIKE é usado para encontrar textos que contenha alguma parte desejada, % significa qualquer coisa 
```

```python
sql(""" SELECT COUNT(*) AS total_vendas,
       SUM(preco * quantidade) AS receita,
       AVG(preco * quantidade) AS media,
       MIN(preco) AS menor_preco,
       MAX(preco) AS maior_preco
FROM vendas """)
```

```python
sql(""" SELECT cidade, categoria, SUM(preco * quantidade) AS receita
FROM vendas
GROUP BY cidade, categoria
ORDER BY cidade, receita DESC """)
```

```python
sql(""" SELECT produto, SUM(quantidade) AS unidades
FROM vendas
WHERE pagamento = 'Pix'
GROUP BY produto
HAVING SUM(quantidade) >= 5
ORDER BY unidades DESC """)
```

```python
sql(""" SELECT COUNT(*) AS n 
FROM vendas 
WHERE pagamento == 'Pix' """)
```

```python
sql(""" SELECT produto, SUM(preco * quantidade) AS receita
FROM vendas GROUP BY produto ORDER BY receita DESC """)
```

```python
sql(""" SELECT produto, SUM(quantidade) AS unidades
FROM vendas GROUP BY produto HAVING SUM(quantidade) > 5
ORDER BY unidades DESC """)
```

```python
sql(""" SELECT categoria, SUM(preco * quantidade) AS receita
FROM vendas WHERE cidade = 'Teresina'
GROUP BY categoria ORDER BY receita DESC """)
```

```python
sql(""" SELECT cidade, COUNT(*) AS n_vendas
FROM vendas GROUP BY cidade HAVING COUNT(*) > 4
ORDER BY n_vendas DESC """)
```
